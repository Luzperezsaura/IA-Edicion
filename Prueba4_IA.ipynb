```python
import yaml

# Estructura de puntos (pi1, pi2, ti1, ti2) = 4 keypoints
data = {
    'path': '/content/dataset_tmRetroPieIzq', # Ruta base absoluta en Colab
    'train': 'images/train',
    'val': 'images/val',
    'names': {0: 'zona_piernas'},
    'kpt_shape': [4, 3] # 4 keypoints, 3 dimensiones (x, y, visibilidad)
}

# Guardamos el archivo
with open('/content/data.yaml', 'w') as f:
    yaml.dump(data, f)

print("Archivo data.yaml creado correctamente.")
```
