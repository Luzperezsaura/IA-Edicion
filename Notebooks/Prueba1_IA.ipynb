Creo archivo de configuración:

```python
import yaml
# Estructura de puntos (pi1, pi2, ti1, ti2) = 4 keypoints
data = {
    'path': '/content/dataset_biomecanica', # Ruta base absoluta en Colab
    'train': 'images/train',
    'val': 'images/val',
    'names': {0: 'zona_piernas'},
    'kpt_shape': [4, 3] # 4 keypoints, 3 dimensiones (x, y, visibilidad)
}

# Guardamos el archivo
with open('/content/data.yaml', 'w') as f:
    yaml.dump(data, f)

print("Archivo data.yaml creado correctamente.")
```

### **Entrenamiento**

```python
# Ejecutamos el entrenamiento por consola (CLI) de Ultralytics
!yolo pose train data=/content/data.yaml model=yolov8n-pose.pt epochs=100 imgsz=640 batch=16 project=biomecanica_proyecto name=modelo_ojalillos device=0
```
